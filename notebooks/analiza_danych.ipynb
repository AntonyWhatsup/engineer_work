```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Loading input data
print("⏳ Loading data...")
# We load the first 100k lines for efficiency
df = pd.read_csv("../data/accepted_2007_to_2018Q4.csv", nrows=100000, low_memory=False)
print("✅ Data loaded successfully!")

# 2. WHAT'S INSIDE? (Data Preview)
print("\n--- PIERWSZE 5 WIERSZY ---")
display(df.head())

# 3. WHAT ARE THE COLUMNS?
print("\n--- LISTA WSZYSTKICH KOLUMN ---")
print(list(df.columns))

# 4. SPRAWDZENIE ZMIENNEJ DOCELOWEJ
print("\n--- CREDIT STATISTICS ---")
print(df['loan_status'].value_counts())

# Chart 1: Status distribution (Raw data)
plt.figure(figsize=(10,6))
sns.countplot(x='loan_status', data=df)
plt.title("Loan Status Distribution (Raw Data)")
plt.xticks(rotation=45)
plt.show()

# 5. CHECK FOR MISSING VALUES
missing = df.isnull().sum()
print("\n--- TOP 10 KOLUMN Z BRAKAMI DANYCH ---")
print(missing[missing > 0].sort_values(ascending=False).head(10))

# --- THIS IS WHERE THE KeyError FIX STARTS ---

print("\n--- PRZYGOTOWANIE DO ANALIZY KORELACJI ---")
# To build a correlation map we need:
# A) Filter only completed loans (to make the analysis reliable)
df_clean = df[df['loan_status'].isin(['Fully Paid', 'Charged Off'])].copy()

# B) Create a numeric column (1 - repaid, 0 - unpaid/debolt)
# This column was missing in the previous version of the analysis
df_clean['loan_status_numeric'] = df_clean['loan_status'].apply(lambda x: 1 if x == 'Fully Paid' else 0)

# Graph 2: Correlation Heat Map (Works on df_clean)
plt.figure(figsize=(10,8))
sns.heatmap(df_clean[['annual_inc', 'loan_amnt', 'fico_range_low', 'loan_status_numeric']].corr(), 
            annot=True, 
            cmap='coolwarm', 
            fmt=".2f")
plt.title("Correlation of variables (What influences loan repayment?)")
plt.show()
```
